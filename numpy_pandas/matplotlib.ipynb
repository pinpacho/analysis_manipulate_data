# Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt


# Grafico de lineas

month = np.array(['E','F','M','A','J'])
sales = np.array([2500,3000,2800,3500,4000])

plt.figure(figsize=(10,6))
plt.plot(month, sales, color='purple', marker='o', linestyle='--')
plt.title('Ventas mensuales de un producto')
plt.xlabel('Meses')
plt.ylabel('Ventas en USD')
plt.show()
```

```python
#Grafico de dispersion es un tipo de grafico que muestra la relacion entre dos variables cuantitativas

#Tamaño del grafico
plt.figure(figsize=(8,6))
hours = [2,3,4,5,6,7,8]
exam_scores = [55,60,65,70,40,80,85]
plt.scatter(hours, exam_scores, color='green', marker='x')
plt.title('Relación entre horas de estudio y puntajes de exámenes')
plt.xlabel('Horas de estudio')
plt.ylabel('Puntajes de exámenes')
plt.show
```

```python
#Grafico de dispersion es un tipo de grafico que muestra la relacion entre dos variables cuantitativas
#De dos estudiantes
#Tamaño del grafico
plt.figure(figsize=(8,6))
hours = [2,3,4,5,6,7,8]
exam_scores_1 = [55,60,65,70,40,80,85]

exam_scores_2 = [53,86,45,70,75,90,100]

#Grafico de dos estudiantes de dispersion

plt.scatter(hours, exam_scores_1, color='green', marker='x',linestyle='-',label='Estudiante 1')
plt.scatter(hours, exam_scores_2, color='blue', marker='s',linestyle='--',label='Estudiante 2')
plt.legend()
plt.title('Relación entre horas de estudio y puntajes de exámenes de dos estudiantes')
plt.xlabel('Horas de estudio')
plt.ylabel('Puntajes de exámenes')
plt.show
```

```python
#Grafico de dispersion es un tipo de grafico que muestra la relacion entre dos variables cuantitativas
#De dos estudiantes
#Tamaño del grafico
plt.figure(figsize=(8,6))
hours = [2,3,4,5,6,7,8]
exam_scores_1 = [55,60,65,70,40,80,85]

exam_scores_2 = [53,86,45,70,75,90,100]

#Grafico de dos estudiantes de dispersion

plt.plot(hours, exam_scores_1, color='green', marker='x',linestyle='-',label='Estudiante 1')
plt.plot(hours, exam_scores_2, color='blue', marker='s',linestyle='--',label='Estudiante 2')
plt.legend()
plt.title('Relación entre horas de estudio y puntajes de exámenes de dos estudiantes')
plt.xlabel('Horas de estudio')
plt.ylabel('Puntajes de exámenes')
plt.show
```

# Barras y Pastel


```python
# Grafico de barras es un tipo de grafico que representa datos categóricos con barras rectangulares cuya altura es proporcional al valor que representan.

plt.figure(figsize=(10,6))
categorias = ['A', 'B', 'C', 'D', 'E']
valores = [23, 45, 56, 78, 12]
plt.bar(categorias, valores, color='red',label='Valores por categoría')

#Destacar un punto específico
plt.annotate('Valor más alto', xy=('D', 78), xytext=('E', 80),
             arrowprops=dict(facecolor='black', shrink=0.05))

plt.legend()
plt.title('Gráfico de Barras de Categorías')
plt.xlabel('Categorías')
plt.ylabel('Valores')
plt.show()
```

```python
#Barras horizontales
plt.figure(figsize=(10,6))
plt.barh(categorias, valores, color='orange',label='Valores por categoría')
plt.legend()
plt.title('Gráfico de Barras de Categorías')
plt.xlabel('Categorías')
plt.ylabel('Valores')
plt.show()
```

```python
# Graficos de pastel son una representación gráfica de datos en forma de un círculo dividido en sectores, donde cada sector representa una proporción del total.
plt.pie(valores, labels=categorias, autopct='%1.1f%%', startangle=140, colors=['gold', 'lightblue', 'lightgreen', 'lightcoral', 'lavender'])
plt.title('Distribución de Valores por Categoría')
plt.show()  
```

# Histogramas y Boxplots

```python
# Los histogramas son una representación gráfica de la distribución de un conjunto de datos continuos, mostrando la frecuencia de los valores en intervalos específicos llamados "bins".

data = np.random.normal(170,10,200)
print(data)
```

```python
plt.figure(figsize=(10,6))
plt.hist(data, bins=20, color='salmon', edgecolor='black', alpha=0.5)
plt.title('Histograma de Alturas')
plt.xlabel('Altura (cm)')
plt.ylabel('Frecuencia')
plt.show()
```

```python
# Los boxplots, también conocidos como diagramas de caja, son una representación gráfica que muestra la distribución de un conjunto de datos a través de sus cuartiles, destacando la mediana, los valores atípicos y la dispersión de los datos.

np.random.seed(0)

ages = [np.random.normal(30,5,100),np.random.normal(40,7,100), np.random.normal(50,6,100)]

```

```python
plt.figure(figsize=(8,6))
plt.boxplot(ages, vert=True, patch_artist=True, notch=True,label=['Grupo 1','Grupo 2','Grupo 3'])
plt.title('Boxplot de Edades por Grupo')
plt.xlabel('Grupos')
plt.ylabel('Edades')
plt.show()
```

# Visualización de Series Temporales

```python
import pandas as pd

# Visualización de Series Temporales

from matplotlib.dates import DateFormatter

dates = pd.date_range(start= '2023-01-01',periods=100)

values = np.random.rand(100).cumsum()

data = pd.DataFrame({'Fecha':dates , 'Valor':values})
```

```python
# Grafico de lineas

fig , ax = plt.subplots(figsize = (12,6))

ax.plot(data['Fecha'], data['Valor'], color='teal', linestyle='-')

plt.xticks(rotation = 45)
plt.title('Serie de Tiempo con formato en las fechas')
plt.xlabel('Fecha')
plt.ylabel('Valor')
```

```python
dates = pd.date_range(start= '2023-01-01',periods=12,freq='M')
sales = np.random.randint(1000,5000,12)

sales_data = pd.DataFrame({'Fecha':dates , 'Ventas':sales})

plt.plot(sales_data['Fecha'],sales_data['Ventas'], marker='o', color='purple', linestyle='-',label='Ventas Mensuales')

plt.gca().xaxis.set_major_formatter(DateFormatter('%b %Y'))

plt.xticks(rotation=45)
plt.title('Ventas Mensuales')
plt.xlabel('Mes')
plt.ylabel('Ventas')
plt.legend()
plt.tight_layout()
plt.show()
```

# Multiples Graficos

```python
import matplotlib.gridspec as gridspec

x = np.linspace(0,10,100)
y1 = np.sin(x)

data = np.random.randn(100)
```

```python
gs = gridspec.GridSpec(2,2,height_ratios=[2,1],width_ratios=[1,1])#Esto hace que la 
#primera fila sea el doble de alta que la segunda y ambas columnas tengan el mismo ancho.

fig = plt.figure(figsize=(10,8))

# Primer Subplot, Ocupa la primera fila

ax1 = fig.add_subplot(gs[0,:])
ax1.plot(x,y1,color = 'blue')
ax1.set_title('Gráfico de Seno')
ax1.set_xlabel('X')
ax1.set_ylabel('sin(X)')

# Segundo subplot, fila 2 columna 1, esquina inferior izquierda

ax2 = fig.add_subplot(gs[1,0])
ax2.hist(data,bins=15,color='green',edgecolor='black',alpha=0.7)
ax2.set_title('Histograma')
ax2.set_xlabel('Valor')
ax2.set_ylabel('Frecuencia')

# Tercer Subplot , esquina inferior derecha

ax3 = fig.add_subplot(gs[1,1])
ax3.scatter(x, y1, color='red', alpha=0.5)
ax3.set_title('Gráfico de Dispersión')
ax3.set_xlabel('X')
ax3.set_ylabel('sin(X)')


plt.tight_layout()
plt.show()
```
